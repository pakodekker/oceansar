# OceanSAR DOpSCa simulation

Notebook equivalent of `oceansar/dopscasim/oceansar_dopscasim.py`. Set the parameter file below, then run all cells.

```python
%load_ext autoreload
%autoreload 2

from pathlib import Path

from oceansar.dopscasim.oceansar_dopscasim import dopscasim
```

## Configuration

Set `cfg_file` to the parameter file to use. The default is `par/dopsca.cfg`.

```python
cfg_file = Path("par/dopsca.cfg")

# Support kernels started either in the repository root or in Notebooks/.
if not cfg_file.exists():
    cfg_file = Path("..") / cfg_file
cfg_file = cfg_file.resolve()

if not cfg_file.is_file():
    raise FileNotFoundError(f"Parameter file not found: {cfg_file}")

cfg_file
```

## Run DOpSCa simulation

The stages executed are controlled by the flags in the `[sim]` section of the parameter file. While `dopsca_raw` is running, its azimuth-cycle progress bar is displayed below the cell.

```python
dopscasim(str(cfg_file))
```
